```python
# ================================================================
# PERSONAL PRODUCTIVITY PATTERN ANALYZER
# COMPLETE EXPLORATORY DATA ANALYSIS (EDA)
# ================================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings

warnings.filterwarnings("ignore")

# ------------------------------------------------
# 1. READ DATASET
# ------------------------------------------------

df = pd.read_csv("/content/train.csv")

print("=" * 70)
print("PERSONAL PRODUCTIVITY PATTERN ANALYZER")
print("EXPLORATORY DATA ANALYSIS")
print("=" * 70)

print("\nDataset loaded successfully!")
print("Rows    :", df.shape[0])
print("Columns :", df.shape[1])

display(df.head())


# ------------------------------------------------
# 2. DISPLAY DATASET INFORMATION
# ------------------------------------------------

print("\n" + "=" * 70)
print("DATASET INFORMATION")
print("=" * 70)

print("\nShape:")
print(df.shape)

print("\nColumns:")
print(df.columns.tolist())

print("\nData Types:")
display(df.dtypes.to_frame("Data Type"))

print("\nDataset Info:")
df.info()


# ------------------------------------------------
# 3. DESCRIPTIVE STATISTICS
# ------------------------------------------------

print("\n" + "=" * 70)
print("DESCRIPTIVE STATISTICS")
print("=" * 70)

display(df.describe().T)


# ------------------------------------------------
# 4. MISSING VALUES
# ------------------------------------------------

print("\n" + "=" * 70)
print("MISSING VALUE ANALYSIS")
print("=" * 70)

missing = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Missing Percentage": (df.isnull().mean() * 100).round(2)
})

display(missing)

print(
    "Total Missing Values:",
    df.isnull().sum().sum()
)


# ------------------------------------------------
# 5. DUPLICATES
# ------------------------------------------------

print("\n" + "=" * 70)
print("DUPLICATE ANALYSIS")
print("=" * 70)

duplicates = df.duplicated().sum()

print("Number of duplicate rows:", duplicates)


# ------------------------------------------------
# 6. NUMERICAL AND CATEGORICAL COLUMNS
# ------------------------------------------------

print("\n" + "=" * 70)
print("FEATURE TYPES")
print("=" * 70)

numerical_columns = df.select_dtypes(
    include=np.number
).columns.tolist()

categorical_columns = df.select_dtypes(
    exclude=np.number
).columns.tolist()

print("\nNumerical Columns:")
print(numerical_columns)

print("\nCategorical Columns:")
print(categorical_columns)


# ------------------------------------------------
# 7. GENDER DISTRIBUTION
# ------------------------------------------------

plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x="gender"
)

plt.title("Gender Distribution")
plt.xlabel("Gender")
plt.ylabel("Number of Students")

plt.tight_layout()
plt.show()


# ------------------------------------------------
# 8. AGE DISTRIBUTION
# ------------------------------------------------

plt.figure(figsize=(8,5))

sns.histplot(
    data=df,
    x="age",
    bins=20,
    kde=True
)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Students")

plt.tight_layout()
plt.show()


# ------------------------------------------------
# 9. PRODUCTIVITY SCORE DISTRIBUTION
# ------------------------------------------------

print("\n" + "=" * 70)
print("PRODUCTIVITY SCORE")
print("=" * 70)

display(
    df["productivity_score"].describe()
)

plt.figure(figsize=(10,5))

sns.histplot(
    data=df,
    x="productivity_score",
    bins=30,
    kde=True
)

plt.title("Distribution of Productivity Score")
plt.xlabel("Productivity Score")
plt.ylabel("Number of Students")

plt.tight_layout()
plt.show()


# ------------------------------------------------
# 10. PRODUCTIVITY BOXPLOT
# ------------------------------------------------

plt.figure(figsize=(10,3))

sns.boxplot(
    x=df["productivity_score"]
)

plt.title("Productivity Score Boxplot")
plt.xlabel("Productivity Score")

plt.tight_layout()
plt.show()


# ------------------------------------------------
# 11. STUDY HOURS DISTRIBUTION
# ------------------------------------------------

plt.figure(figsize=(9,5))

sns.histplot(
    data=df,
    x="study_hours_per_day",
    bins=25,
    kde=True
)

plt.title("Distribution of Study Hours per Day")
plt.xlabel("Study Hours per Day")
plt.ylabel("Number of Students")

plt.tight_layout()
plt.show()


# ------------------------------------------------
# 12. SLEEP HOURS DISTRIBUTION
# ------------------------------------------------

plt.figure(figsize=(9,5))

sns.histplot(
    data=df,
    x="sleep_hours",
    bins=25,
    kde=True
)

plt.title("Distribution of Sleep Hours")
plt.xlabel("Sleep Hours")
plt.ylabel("Number of Students")

plt.tight_layout()
plt.show()


# ------------------------------------------------
# 13. PHONE USAGE DISTRIBUTION
# ------------------------------------------------

plt.figure(figsize=(9,5))

sns.histplot(
    data=df,
    x="phone_usage_hours",
    bins=25,
    kde=True
)

plt.title("Distribution of Phone Usage Hours")
plt.xlabel("Phone Usage Hours")
plt.ylabel("Number of Students")

plt.tight_layout()
plt.show()


# ------------------------------------------------
# 14. DIGITAL DISTRACTION FEATURES
# ------------------------------------------------

distraction_columns = [
    "social_media_hours",
    "youtube_hours",
    "gaming_hours"
]

fig, axes = plt.subplots(
    1, 3,
    figsize=(18,5)
)

for ax, column in zip(
    axes,
    distraction_columns
):

    sns.histplot(
        data=df,
        x=column,
        bins=20,
        kde=True,
        ax=ax
    )

    ax.set_title(
        column.replace("_", " ").title()
    )

plt.tight_layout()
plt.show()


# ------------------------------------------------
# 15. EXERCISE AND BREAKS
# ------------------------------------------------

fig, axes = plt.subplots(
    1, 2,
    figsize=(14,5)
)

sns.histplot(
    data=df,
    x="exercise_minutes",
    bins=25,
    kde=True,
    ax=axes[0]
)

axes[0].set_title(
    "Exercise Minutes Distribution"
)

sns.histplot(
    data=df,
    x="breaks_per_day",
    bins=15,
    kde=True,
    ax=axes[1]
)

axes[1].set_title(
    "Breaks per Day Distribution"
)

plt.tight_layout()
plt.show()


# ------------------------------------------------
# 16. STUDY HOURS VS PRODUCTIVITY
# ------------------------------------------------

plt.figure(figsize=(9,6))

sns.scatterplot(
    data=df,
    x="study_hours_per_day",
    y="productivity_score",
    alpha=0.4
)

sns.regplot(
    data=df,
    x="study_hours_per_day",
    y="productivity_score",
    scatter=False,
    ci=None
)

plt.title("Study Hours vs Productivity Score")
plt.xlabel("Study Hours per Day")
plt.ylabel("Productivity Score")

plt.tight_layout()
plt.show()


# ------------------------------------------------
# 17. SLEEP VS PRODUCTIVITY
# ------------------------------------------------

plt.figure(figsize=(9,6))

sns.scatterplot(
    data=df,
    x="sleep_hours",
    y="productivity_score",
    alpha=0.4
)

sns.regplot(
    data=df,
    x="sleep_hours",
    y="productivity_score",
    scatter=False,
    ci=None
)

plt.title("Sleep Hours vs Productivity Score")
plt.xlabel("Sleep Hours")
plt.ylabel("Productivity Score")

plt.tight_layout()
plt.show()


# ------------------------------------------------
# 18. PHONE USAGE VS PRODUCTIVITY
# ------------------------------------------------

plt.figure(figsize=(9,6))

sns.scatterplot(
    data=df,
    x="phone_usage_hours",
    y="productivity_score",
    alpha=0.4
)

sns.regplot(
    data=df,
    x="phone_usage_hours",
    y="productivity_score",
    scatter=False,
    ci=None
)

plt.title("Phone Usage vs Productivity Score")
plt.xlabel("Phone Usage Hours")
plt.ylabel("Productivity Score")

plt.tight_layout()
plt.show()


# ------------------------------------------------
# 19. EXERCISE VS PRODUCTIVITY
# ------------------------------------------------

plt.figure(figsize=(9,6))

sns.scatterplot(
    data=df,
    x="exercise_minutes",
    y="productivity_score",
    alpha=0.4
)

plt.title("Exercise Minutes vs Productivity Score")
plt.xlabel("Exercise Minutes")
plt.ylabel("Productivity Score")

plt.tight_layout()
plt.show()


# ------------------------------------------------
# 20. FOCUS VS PRODUCTIVITY
# ------------------------------------------------

plt.figure(figsize=(9,6))

sns.scatterplot(
    data=df,
    x="focus_score",
    y="productivity_score",
    alpha=0.4
)

sns.regplot(
    data=df,
    x="focus_score",
    y="productivity_score",
    scatter=False,
    ci=None
)

plt.title("Focus Score vs Productivity Score")
plt.xlabel("Focus Score")
plt.ylabel("Productivity Score")

plt.tight_layout()
plt.show()


# ------------------------------------------------
# 21. STRESS VS PRODUCTIVITY
# ------------------------------------------------

plt.figure(figsize=(9,6))

sns.boxplot(
    data=df,
    x="stress_level",
    y="productivity_score"
)

plt.title("Stress Level vs Productivity Score")
plt.xlabel("Stress Level")
plt.ylabel("Productivity Score")

plt.tight_layout()
plt.show()


# ------------------------------------------------
# 22. ASSIGNMENTS VS PRODUCTIVITY
# ------------------------------------------------

plt.figure(figsize=(9,6))

sns.scatterplot(
    data=df,
    x="assignments_completed",
    y="productivity_score",
    alpha=0.4
)

plt.title("Assignments Completed vs Productivity Score")
plt.xlabel("Assignments Completed")
plt.ylabel("Productivity Score")

plt.tight_layout()
plt.show()


# ------------------------------------------------
# 23. ATTENDANCE VS PRODUCTIVITY
# ------------------------------------------------

plt.figure(figsize=(9,6))

sns.scatterplot(
    data=df,
    x="attendance_percentage",
    y="productivity_score",
    alpha=0.4
)

plt.title("Attendance Percentage vs Productivity Score")
plt.xlabel("Attendance Percentage")
plt.ylabel("Productivity Score")

plt.tight_layout()
plt.show()


# ------------------------------------------------
# 24. FINAL GRADE VS PRODUCTIVITY
# ------------------------------------------------

plt.figure(figsize=(9,6))

sns.scatterplot(
    data=df,
    x="final_grade",
    y="productivity_score",
    alpha=0.4
)

plt.title("Final Grade vs Productivity Score")
plt.xlabel("Final Grade")
plt.ylabel("Productivity Score")

plt.tight_layout()
plt.show()


# ------------------------------------------------
# 25. CORRELATION HEATMAP
# ------------------------------------------------

numeric_df = df.select_dtypes(
    include=np.number
)

correlation_matrix = numeric_df.corr()

plt.figure(
    figsize=(16,12)
)

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    linewidths=0.3
)

plt.title(
    "Correlation Heatmap",
    fontsize=16
)

plt.tight_layout()
plt.show()


# ------------------------------------------------
# 26. CORRELATION WITH PRODUCTIVITY
# ------------------------------------------------

print("\n" + "=" * 70)
print("CORRELATION WITH PRODUCTIVITY SCORE")
print("=" * 70)

productivity_correlation = (
    numeric_df
    .corr()["productivity_score"]
    .sort_values(ascending=False)
)

display(
    productivity_correlation.to_frame(
        name="Correlation"
    )
)


# ------------------------------------------------
# 27. PRODUCTIVITY CORRELATION BAR CHART
# ------------------------------------------------

corr_features = (
    productivity_correlation
    .drop("productivity_score")
    .sort_values()
)

plt.figure(figsize=(10,8))

sns.barplot(
    x=corr_features.values,
    y=corr_features.index
)

plt.title(
    "Feature Correlation with Productivity Score"
)

plt.xlabel("Correlation")
plt.ylabel("Feature")

plt.tight_layout()
plt.show()


# ------------------------------------------------
# 28. OUTLIER ANALYSIS
# ------------------------------------------------

outlier_columns = [
    "study_hours_per_day",
    "sleep_hours",
    "phone_usage_hours",
    "social_media_hours",
    "youtube_hours",
    "gaming_hours",
    "breaks_per_day",
    "coffee_intake_mg",
    "exercise_minutes",
    "assignments_completed",
    "attendance_percentage",
    "stress_level",
    "focus_score",
    "final_grade",
    "productivity_score"
]

fig, axes = plt.subplots(
    4,
    4,
    figsize=(16,18)
)

axes = axes.flatten()

for i, column in enumerate(
    outlier_columns
):

    sns.boxplot(
        x=df[column],
        ax=axes[i]
    )

    axes[i].set_title(
        column.replace("_", " ").title()
    )

for i in range(
    len(outlier_columns),
    len(axes)
):
    axes[i].axis("off")

plt.tight_layout()
plt.show()


# ------------------------------------------------
# 29. PAIRPLOT
# ------------------------------------------------

pairplot_columns = [
    "study_hours_per_day",
    "sleep_hours",
    "phone_usage_hours",
    "stress_level",
    "focus_score",
    "productivity_score"
]

pairplot_data = df[
    pairplot_columns
].sample(
    min(2500, len(df)),
    random_state=42
)

sns.pairplot(
    pairplot_data,
    corner=True,
    diag_kind="hist"
)

plt.show()


# ------------------------------------------------
# 30. PRODUCTIVITY BY GENDER
# ------------------------------------------------

gender_productivity = (
    df
    .groupby("gender")["productivity_score"]
    .agg([
        "count",
        "mean",
        "median",
        "std"
    ])
)

print("\nProductivity Statistics by Gender:")
display(gender_productivity)

plt.figure(figsize=(8,5))

sns.boxplot(
    data=df,
    x="gender",
    y="productivity_score"
)

plt.title("Productivity Score by Gender")
plt.xlabel("Gender")
plt.ylabel("Productivity Score")

plt.tight_layout()
plt.show()


# ------------------------------------------------
# 31. FINAL EDA SUMMARY
# ------------------------------------------------

print("\n" + "=" * 70)
print("FINAL EDA SUMMARY")
print("=" * 70)

print("\nDATASET")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

print("\nDATA QUALITY")
print(
    "Missing values:",
    df.isnull().sum().sum()
)

print(
    "Duplicate rows:",
    df.duplicated().sum()
)

print("\nPRODUCTIVITY SCORE")

print(
    "Minimum:",
    round(
        df["productivity_score"].min(),
        2
    )
)

print(
    "Maximum:",
    round(
        df["productivity_score"].max(),
        2
    )
)

print(
    "Mean:",
    round(
        df["productivity_score"].mean(),
        2
    )
)

print(
    "Median:",
    round(
        df["productivity_score"].median(),
        2
    )
)


# Correlations
corr = (
    numeric_df
    .corr()["productivity_score"]
    .drop("productivity_score")
    .sort_values(
        ascending=False
    )
)

print("\nTOP POSITIVE CORRELATIONS")

display(
    corr.head(5).to_frame(
        "Correlation"
    )
)

print("\nTOP NEGATIVE CORRELATIONS")

display(
    corr.tail(5)
    .sort_values()
    .to_frame(
        "Correlation"
    )
)


# ------------------------------------------------
# 32. COMPLETION MESSAGE
# ------------------------------------------------

print("\n" + "=" * 70)
print("EDA COMPLETED SUCCESSFULLY!")
print("=" * 70)

```
